```python
def calculator(hours, days):
    return hours * days

def study_plan(subject, days, total_hours):
    hours_per_day = total_hours / days

    plan = f"\nStudy Plan for {subject}\n\n"

    topics = [
        "Basics",
        "Important Concepts",
        "Practical Examples",
        "Practice Questions",
        "Revision"
    ]

    # Generate only the number of days requested
    for day in range(1, days + 1):

        # If days are more than 5, repeat the last topic
        if day <= len(topics):
            topic = topics[day - 1]
        else:
            topic = "Practice & Revision"

        plan += f"Day {day}: {topic}\n"

    plan += f"\nStudy {hours_per_day:.1f} hours per day."

    return plan


def agent(subject, days, hours_per_day):

    print("🤖 Agent is thinking...")

    # Agent decides to use calculator
    print("🔧 Using Calculator Tool...")
    total_hours = calculator(hours_per_day, days)

    # Agent decides to use planner
    print("📝 Using Study Planner Tool...")
    result = study_plan(subject, days, total_hours)

    return result


# User Input
subject = input("What do you want to study? ")
days = int(input("How many days do you have? "))
hours = float(input("How many hours can you study per day? "))

# Agent
answer = agent(subject, days, hours)

print(answer)
```
